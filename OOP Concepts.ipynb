```python
class Bike:
    color = 'black'
    model = 125
    def drive(self):
        print(f'{self.color}, and {self.model} Bike is riding')
bike_1 = Bike()
bike_1.drive()
```

```python
class Bike:
    color = 'Blue'
    def drive(self):
        print(f'{self.color} Bike is riding!')
    def set_color(self, new_color):
        self.color = new_color
bike_1 = Bike()
bike_1.set_color('Red')
bike_1.drive()
```

```python
class Cat:
    def __init__(self):
        self.name = 'puppy'
    def barke(self):
        print(f'{self.name} is saying Meoww!')
cat_1 = Cat()
cat_1.barke()
```

```python
class Bike:
    def __init__ (self,brand,color,engine):
        self.brand = brand
        self.color = color
        self.engine = '125cc'
bike_1 = Bike('Yahama','Black','125cc')
print(bike_1.brand)
print(bike_1.color)
print(bike_1.engine)
```

```python
class Students:
    def __init__ (self,name,age,agno):
        self.name = name
        self.age = age
        self.agno = agno
    def display_info(self):
        print(f'Student name: {self.name}')
        print(f'Student age: {self.age}')
        print(f'student agno: {self.agno}')
    def is_eligible(self):
        if self.age >=18:
            print(self.name, 'is eligible for admission.')
        else:
            print(self.name, 'is not eligible for admission.')
std_1 = Students('Hamza',22,'2023-ag-8720')
std_2 = Students('Qadeer',17,'2023-ag-8698')
print(std_1.name)
print(std_1.age)
print(std_1.agno)
print(std_2.name)
print(std_2.age)
print(std_2.agno)
```

```python
std_1.display_info()
std_1.is_eligible()
```

```python
std_2.display_info()
std_2.is_eligible()
```
